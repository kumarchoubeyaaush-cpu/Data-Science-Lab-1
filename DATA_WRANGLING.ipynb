```python
A = pd.DataFrame({
                 "Name":["Aayush","Priya"],
                 "Marks":[99,89]
})
B = pd.DataFrame({
                   "Name":["Aman","Riya"],
                 "Marks":[34,59]
})
c = pd.concat([A,B])
print(c)
```

```python
A = pd.DataFrame({

                 "Name":["Aayush","Priya"],
                 "Marks":[99,89]
})
B = pd.DataFrame({

                   "Name":["Aman","Riya"],
                 "Marks":[34,59]
})
c = pd.concat([A,B],ignore_index=True)
print(c)
```

```python
# concat columns wise
A1 = pd.DataFrame({
                 "Name":["Aayush","Priya"],
                 "Marks":[99,89]
})
B1 = pd.DataFrame({
                   "Attendance":[99,80]
})
C1 = pd.concat([A1,B1],axis = 1)
print(C1)
```

```python
# inner join
Student = pd.DataFrame({
                 "Student_id":[101,102,103,104],
                 "Name":["Aayush","Priya","Raj","Riya"]
})
Marks = pd.DataFrame({
                  "Student_id":[101,102,103,104],
                  "Marks":[99,89,90,70]
})
result=pd.merge(Student,Marks,on = "Student_id")
print(result)
```

```python
Student = pd.DataFrame({
                 "Student_id":[101,102,103,104],
                 "Name":["Aayush","Priya","Raj","Riya"]
})
Marks = pd.DataFrame({
                  "Student_id":[101,102,103,105],
                  "Marks":[99,89,90,70]
})
result=pd.merge(Student,Marks,on = "Student_id")
print(result)
```

```python
# inner join
pd.merge(Student,Marks,on="Student_id",how="inner")
```

```python
# left join
pd.merge(Student,Marks,on="Student_id",how="left")
```

```python
#right join
pd.merge(Student,Marks,on="Student_id",how="right")
```

```python
#outer join
pd.merge(Student,Marks,on="Student_id",how="outer")
```

```python
#cross join
left = pd.DataFrame({"Student":["Aayush","Priya"]})
right = pd.DataFrame({"Subject":["JAVA","C++"]})
pd.merge(left,right,how="cross")
```

```python
df2 = pd.DataFrame({"Student_id":[101,102],
                    "Marks":[33,49]
                    })
df3 = pd.DataFrame({"Student_id":[101,102],
                    "Marks":[45,79]
                    })
result = pd.merge(df2,df3,on="Student_id", suffixes=("_internal","_external"))
print(result)
```

```python
r1=pd.merge(Student,Marks,on="Student_id",how = "outer",indicator="Source")
print(r1)
```

```python
# one_to_one
Student = pd.DataFrame({
                 "Student_id":[101,102,103,104],
                 "Name":["Aayush","Priya","Raj","Riya"]
})
Marks = pd.DataFrame({
                  "Student_id":[101],
                  "Marks":[99]
})
pd.merge(Student,Marks,on="Student_id",validate="one_to_one")
```

```python
#  one_to_many

Student = pd.DataFrame({
                 "Student_id":[101,102,103,104],
                 "Name":["Aayush","Priya","Raj","Riya"]
})
Marks = pd.DataFrame({
                  "Student_id":[101,101],
                  "Marks":[99,88]
})
pd.merge(Student,Marks,on="Student_id",validate="one_to_many")
```

```python
df= pd.DataFrame({
                   "Student":["Amit","Amit","Reena","Reena","Raj","Raj"],
                   "Subject":["Python","Java","Python","Java","Python","Java"],
                   "Marks":[80,70,45,66,77,90]
})
print(df)
```

```python
res = df.pivot(index = "Student",columns="Subject",values="Marks")
print(res)
```
