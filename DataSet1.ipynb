```python
# MEAN ,MEDIAN , MODE
cols = df.select_dtypes(include=['number']).columns

mean = df[cols].mean()
median = df[cols].median()
mode = df[cols].mode().iloc[0]

print("Mean:")
print(mean)

print("\nMedian:")
print(median)

print("\nMode:")
print(mode)
```

```python
# CONVERT ZEROES TO NUN
import numpy as np

cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

df[cols] = df[cols].replace(0, np.nan)
```

```python
cols = df.select_dtypes(include=['number']).columns

mean = df[cols].mean()
median = df[cols].median()
mode = df[cols].mode().iloc[0]

print("Mean:")
print(mean)

print("\nMedian:")
print(median)

print("\nMode:")
print(mode)
```
