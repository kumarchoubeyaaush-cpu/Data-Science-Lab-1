```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


```

```python
from sklearn.model_selection import train_test_split
```

```python
# CREATE A DATASET : 100 student record
np.random.seed(42)
```

```python
df = pd.DataFrame({
    'student_ID':range(1,101),
    'Name':[f'Student_{i}'for i in range(1,101)],
    'Gender':np.random.choice(['Female' , 'Male'],100),
    'Department':np.random.choice(['CSE' , 'ECE' , 'ME' , 'MATHEMATICS'],100),
    'Marks':np.random.randint(40,101,100)
})
```

```python
df
```

```python
# Population Size
print(len(df))
print("Population Size :",len(df))

#About Data/Population Ispection
print("Department Distribution",df['Department'].value_counts())

#Gender Distribution
print("Gender Distribution",df['Gender'].value_counts())

#Marks Distribution
print("\n Basic Distribution of Marks:",df['Marks'].describe())
```

```python
# Simple Random Sampling : Equal Chance for selection of every population
sample_size = 10
simple_random_sample = df.sample(
     n=sample_size,
     random_state=42
 )
```

```python
simple_random_sample
```

```python
# VISUALIZATION for simple Random selection
plt.figure(figsize=(12,6))

plt.scatter(df['student_ID'],df['Marks'],alpha=0.35,label='Population')

plt.scatter(simple_random_sample['student_ID'],
            simple_random_sample['Marks'],
            s=120,label='Selected Sample'
            )
plt.xlabel("STUDENT ID")
plt.ylabel("MARKS")
plt.title("SIMPLE RAMDOM SAMPLING")
plt.legend()
plt.show()
```

SYSTEMATIC SAMPLING

```python
#SYSTEMATIC SAMPLING
N = len(df)
n=10
K = N//n
start = np.random.randint(0,K)
systematci_sampling = df.iloc[start : : K].head(n)
print("Sampling Interval : K=",K)
print("Starting position : ",start)
systematci_sampling
```

```python
#visualization of systematic sampling
plt.figure(figsize=(12,5))
plt.scatter(df["student_ID"], df['Marks'] , alpha=0.35,label = 'population')
plt.scatter(systematci_sampling['student_ID'],systematci_sampling['Marks'],
            s =120, label='Selected Sample')
plt.xlabel("STUDENT_ID")
plt.ylabel("MARKS")
plt.title("Systematic Sampling")
plt.legend()
plt.show
```

```python
# STRATIFIED SAMPLING  : STRATEGY

sample_size=20
stratified_sample , remaning = train_test_split(
    df,
    test_size=0.80,
    stratify=df['Department'],
    random_state=42
)
stratified_sample
```

```python
# CLUSTER SAMPLING

cluster = df['Department'].unique()
selected_clusters = np.random.choice(
    cluster,
    size=2,
    replace=False
)
cluster_sample = df[df['Department'].isin(selected_clusters)]
print("Selected CLuster: ",selected_clusters)
print("\n Sample_size:", len(cluster_sample))
print(cluster_sample)
```

```python
# QUOTA SAMPLING :

#  Quota sampling is a non-probability sampling method where researchers divide
#  a population into mutually exclusive subgroups and recruit participants until specific,
#  predetermined targets (quotas) for each subgroup are met.

female_sample =df[
     df['Gender'] == 'Female'].sample (n =10, random_state = 42)

male_sample = df[
    df['Gender'] == 'Male'].sample ( n=10, random_state = 42)

quota_sample  = pd.concat([
    male_sample, female_sample
])
print(quota_sample['Gender'])
```
