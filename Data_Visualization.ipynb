Using Matplotlib Diagram

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df1=pd.read_csv("/content/sharma-kohli.csv")
df1
```

```python
plt.plot(df1['index'], df1['V Kohli'], label="Virat kohli")
plt.show()
```

```python
plt.plot(df1['index'], df1['V Kohli'])
plt.plot(df1['index'], df1['RG Sharma'])
plt.show()
```

```python
plt.plot(df1['index'], df1['V Kohli'], label="King Kohli")
plt.plot(df1['index'], df1['RG Sharma'], label="Hit Man Sharma")
plt.legend()
plt.show()
```

```python
plt.plot(df1['index'], df1['V Kohli'], label="King Kohli", color='red', marker='o', ls='-',
                                       linewidth=2, markersize=10)

plt.plot(df1['index'], df1['RG Sharma'], label="Hit Man Sharma", color='blue', marker='s', linestyle='--',
                                         linewidth=2, markersize=10)

plt.xlabel('Years',fontsize=30)
plt.ylabel("Runs")
plt.title('Rohit Sharma vs Virat Kohli')
plt.legend(loc='best', fontsize=10)
plt.ylim(100,1500)
plt.grid(True,linestyle=":",alpha=0.9)
#plt.figure(figsize=(10,50))
plt.savefig("IPL.png")
plt.show()
```

```python
India=pd.DataFrame({
    "Score":[24,20,36,42,30],
    "Over":[1,2,3,4,5]
})
Pakistan=pd.DataFrame({
    "Score":[12,19,16,22,14],
    "Over":[1,2,3,4,5]
})

plt.plot('Over','Score',data=India,label='India',color='blue',marker='o')
plt.plot('Over','Score',data=Pakistan,label='Pakistan',color='green',marker='*')
plt.xlabel('Over',fontsize=16)
plt.ylabel('Runs Scored',fontsize=16)
plt.title('India VS Pakistan',fontsize=15)
plt.ylim(1,50)
plt.grid(True,linestyle='--',alpha=0.25)
plt.legend(loc='best')
plt.show()
```
